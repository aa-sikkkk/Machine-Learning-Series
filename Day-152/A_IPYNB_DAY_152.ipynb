### The Concept: Self-Attention
Transformers don't read left-to-right. They read everything at once.
To understand a word, the model "attends" to every other word in the sentence, assigning a relevance score.

**The Mathematical Core:**
$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

-   **Query ($Q$):** The word asking for context (e.g., "bank" in "river bank").
-   **Key ($K$):** The word offering context (e.g., "river").
-   **Value ($V$):** The actual content of the word.
-   **Score:** The dot product $QK^T$ determines how much "attention" is paid.

### The Objective
We will not just run inference. We will visualize the internal wiring of `bert-base-uncased`.
We will answer:
1.  **Syntactic Attention:** Does the model know that adjectives modify nouns?
2.  **Coreference Resolution:** When we say "it", does the model look at the object "it" refers to?

```python
from transformers import AutoTokenizer, AutoModel, utils

# We use 'bert-base-uncased', the standard workhorse of NLP
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name, output_attentions=True)

print(f"Loaded {model_name}. Ready to dissect.")
```

### Visualization 1: The Head View
BERT Base has 12 layers, and each layer has 12 "attention heads". That's 144 distinct mechanisms viewing the text.

**How to interact with the plot below:**
-   **Hover** over a token on the left/right to see incoming/outgoing attention.
-   **Click** on the colored squares at the top to toggle specific attention heads on/off.
-   **Lines:** Thicker lines = Stronger attention (higher softmax score).

**Test Sentence:**
*"The animal didn't cross the street because it was too tired."*
(This is a famous linguistic test. Does "it" refer to "animal" or "street"?)

```python
from bertviz import head_view

# This is our input
sentence = "The animal didn't cross the street because it was too tired."
inputs = tokenizer.encode(sentence, return_tensors='pt')

outputs = model(inputs)
attention = outputs[-1]  # The last element of the output tuple is attentions
tokens = tokenizer.convert_ids_to_tokens(inputs[0])

print("Hover over words to see attention lines. Click colored tiles to filter heads.")
head_view(attention, tokens)
```

### What are you seeing?

If you explore the layers (using the dropdown in the visualization), you will find specific patterns emerge in different heads:

1.  **The "Next Word" Head:**
    * *Where:* Early layers (Layer 1-2).
    * *Behavior:* Most attention lines point to the immediate next word. The model is learning basic n-gram structure (like a Convolutional filter).

2.  **The "Separator" Head:**
    * *Where:* Mid-to-Late layers.
    * *Behavior:* You will see massive attention focused on `[SEP]` or `[CLS]`.
    * *Why:* This is a "no-op". When a head doesn't find anything useful in the sentence, it "parks" its attention on these special tokens to avoid disrupting the signal.

3.  **The "Coreference" Head (The Magic):**
    * *Where:* Layer 9 or 10 usually contains semantic links.
    * *Task:* Hover over the word **"it"**.
    * *Observation:* Look for a head that strongly connects **"it"** back to **"animal"**.
    * *Experiment:* Change the sentence to *"because it was too wide"*. Run it again. The attention should shift from "animal" to "street".

### Visualization 2: The Model View
This provides a bird's-eye view of all 144 attention heads at once.

-   **Rows:** Attention heads.
-   **Columns:** Layers (0 to 11).
-   **Network:** The lines show the attention structure.

**Look for:**
-   **Grid Patterns:** Heads that attend to the current position (self-preservation).
-   **Diagonals:** Heads that look at the previous/next word.
-   **Vertical Columns:** Heads that fixate on a specific token (usually `[SEP]`).

```python
from bertviz import model_view
model_view(attention, tokens)
```

### Visualization 3: The Neuron View (Q, K, V)
This is the deepest level. It visualizes the math directly.

$$\text{Score} = Q \times K$$

-   **Left Column:** Query Vectors (what the word is looking for).
-   **Middle Column:** Key Vectors (what the word offers).
-   **Lines:** The dot product (similarity) between them.

*Note: This view is dense. Select Layer 10, Head 5 (or similar) to see semantic matching.*

```python
from bertviz.transformers_neuron_view import BertModel, BertTokenizer
from bertviz.neuron_view import show
model_type = 'bert'
model_version = 'bert-base-uncased'
model = BertModel.from_pretrained(model_version, output_attentions=True)
tokenizer = BertTokenizer.from_pretrained(model_version, do_lower_case=True)

sentence = "The animal didn't cross the street because it was too tired."
show(model, model_type, tokenizer, sentence, display_mode='light', layer=0, head=0)
```
